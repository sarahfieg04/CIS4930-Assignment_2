```python
intensity = pd.read_excel("raw/Results.xlsx", "Table 3", skiprows=1)
intensity.tail(10)
```

```python
def huc_code(col):
    v = pd.to_numeric(col, errors="coerce")
    return v.map(lambda x: f"{int(x):08d}" if pd.notna(x) else None)

intensity["huc8"] = huc_code(intensity["HUC8"])
intensity = intensity.dropna(subset=["huc8"])
print(len(intensity))
intensity.head()
```
